## 3. Penjelasan Kegunaan Masing-Masing Library

### 3.1 PyPREP
**PyPREP** adalah implementasi Python dari pipeline **PREP (Preprocessing Pipeline)** yang aslinya dikembangkan di MATLAB, digunakan khusus untuk **preprocessing data EEG (Electroencephalogram)**. Library ini bekerja di atas **MNE-Python** dan berfungsi untuk:
- Mendeteksi dan menandai *bad channels* (kanal elektroda yang rusak/noise) secara otomatis.
- Melakukan *referencing* ulang (robust average referencing) pada data EEG.
- Menghilangkan interferensi *line noise* (misalnya noise 50/60 Hz dari listrik AC).

Singkatnya, PyPREP digunakan untuk membersihkan data EEG mentah sebelum dianalisis lebih lanjut, sehingga hasil analisis menjadi lebih akurat dan bebas dari artefak/noise teknis.

### 3.2 SciPy
**SciPy** adalah library ilmiah (*scientific computing*) inti dalam ekosistem Python, dibangun di atas **NumPy**. SciPy menyediakan berbagai modul untuk:
- Pemrosesan sinyal (`scipy.signal`) — filtering, konvolusi, transformasi Fourier, dsb. Sangat berguna untuk memproses sinyal biomedis seperti EEG dan ECG.
- Statistik (`scipy.stats`).
- Optimisasi (`scipy.optimize`).
- Aljabar linear (`scipy.linalg`).
- Interpolasi, integrasi numerik, dan pengolahan citra.

Dalam konteks pengolahan sinyal biomedis, SciPy umumnya digunakan untuk melakukan *filtering* (misalnya bandpass/notch filter) pada sinyal EEG/ECG sebelum diekstraksi fitur.

### 3.3 wandb (Weights & Biases)
**wandb** adalah platform/tool untuk **melacak (tracking) dan memvisualisasikan eksperimen machine learning**. Fungsinya antara lain:
- Mencatat metrik training (loss, accuracy, dll.) secara real-time dan menampilkannya dalam dashboard interaktif.
- Menyimpan (log) hyperparameter, konfigurasi model, serta artefak (model, dataset) untuk keperluan reproduksibilitas.
- Membandingkan hasil dari berbagai eksperimen/model secara mudah.
- Melakukan *hyperparameter tuning* otomatis (wandb Sweeps).

Library ini sangat membantu ketika kita melatih model machine learning/deep learning untuk klasifikasi sinyal EEG/ECG, karena mempermudah pemantauan proses training dan perbandingan antar model.

### 3.4 pyECG
**pyECG** adalah library Python yang digunakan untuk membaca dan menulis (*read/write*) berkas data **ECG (Electrocardiogram)** dalam berbagai format standar, seperti:
- **WFDB** (format Physiobank/PhysioNet, `.hea`/`.dat`)
- **iShine Holter ECG** (`.ecg`/`.ann`)

Dengan pyECG, pengguna dapat memuat rekaman sinyal ECG dari berbagai format dataset publik menjadi objek Python (`ECGRecord`) yang mudah diakses, sehingga dapat langsung diproses, difilter, atau dianalisis lebih lanjut (misalnya untuk deteksi *heartbeat*, analisis *heart rate variability*, atau klasifikasi aritmia).
